```python

class Person:
    def __init__(self, name):
        print("Person created")
        self.name = name
    def print_person(self):
        print(f"Person name= {self.name}")


class Teacher(Person):
    def __init__(self, salary):
        self.salary = salary
        print("Teacher created")


t = Teacher("Abbass")
print(t.__dict__)
t.print_person()
```

```python

class Person:
    def __init__(self, name):
        print("Person created")
        self.name = name
    def print_person(self):
        print(f"Person name= {self.name}")


class Teacher(Person):
    def __init__(self, name, salary):
        self.salary = salary
        self.name = name  # don't do that
        print("Teacher created")


t = Teacher("Abbass", 545)
print(t.__dict__)
t.print_person()
```

```python

class Person:
    def __init__(self, name):
        print("Person created")
        self.name = name
    def print_person(self):
        print(f"Person name= {self.name}")


class Teacher(Person):
    def __init__(self, name, salary):
        # we need to call parent constructor
        super().__init__(name.title())
        self.salary = salary
        print("Teacher created")


t = Teacher("Abbass ahmed", 545)
print(t.__dict__)
t.print_person()
```
