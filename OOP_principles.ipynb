```python
info = [1, "noha", "iti"]

emp= {
    "id":1,
    "name":"noha",
    "salary" : 1000
}

emp2 = {
    "id":2,
    "name":"omar",
    "salary" : 1000
}
```

```python
def print_emp(emp: dict):
    print(f"Employee info name{emp['name']} salary{emp['salary']}, id{emp['id']}")
```

```python
print_emp(emp)
print_emp(emp2)
```

```python
emp3= {
    "emp_id":1,
    "name":"noha",
    "salary" : 1000
}

print_emp(emp3)
```

<span style="font-size:20px; font-weight:700;">How the object is created</span>
<div style="margin-top:8px; font-size:15px;"><code>__init__</code> runs the moment you write <code>Student()</code>. Python passes the brand-new object in as the first argument.</div>
</div>

<img src="illustrations/oop-self-arrow.jpg" alt="A tag labeled self with an arrow pointing at an Employee object that has name and salary. A light bulb marks __init__ as the constructor." style="width:100%; border-radius:16px; margin-top:12px;">

```python
class Student:
    def __init__(self):
        # address saved in self
        print(f"student object is created {self}")
```

```python
s = Student()
```

```python
class Student:
    def __init__():
        # address saved in self
        print("الجو منعش وجميل ")
```

```python
salma =  Student()
print(salma)
```

<div style="background-color:#FDE7E7; border:2px solid #B83232; border-radius:12px; padding:14px 18px; color:#111111;">
<div style="font-size:18px; font-weight:700; color:#B83232;">Please don't do that</div>
<div style="margin-top:8px; font-size:15px;">Leaving out <code>self</code> crashes: Python still sends the new object, and <code>__init__</code> has nowhere to put it.</div>
<div style="margin-top:8px; font-size:15px;">Naming that parameter <code>mine</code> works. Readers still expect <code>self</code>. Keep the usual name.</div>
</div>

```python
class Employee:
    def __init__(mine): # Usually first parameter of a method is named 'self'
        print(f'Employee is created {mine}')

emp = Employee()  #
```

```python
class Employee:
    def __init__(self): # Usually first parameter of a method is named 'self'
        print(f'Employee is created {self}')

emp = Employee()  #  emp ===> refer to object address
print(emp)
```
