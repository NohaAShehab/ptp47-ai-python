<span style="font-size:22px; font-weight:700;">Trim the ends</span>
<div style="margin-top:6px; color:#333333;">strip removes characters from the start and the end only.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #C9491A; border-radius:14px; padding:8px; margin-top:10px;">
<img src="illustrations/strip.svg" alt="strip removing spaces, then stopping at a hash, then removing the hash too" style="width:100%; height:auto; display:block; border-radius:10px;">

```python
bio = "     I love reading much     "
print(len(bio))

bio2 = bio.strip()  # strip white spaces from the beginning and the end of the string
print(bio2, len(bio2))
```

```python
bio = "     I love reading much     #"
print(len(bio))

bio2 = bio.strip()  # strip white spaces from the beginning and the end of the string
print(bio2, len(bio2))
```

```python
bio = "     I love reading much     #"
print(len(bio))

bio2 = bio.strip("#$ ")  # strip any set of chars from beginning and end
print(bio2)
print(len(bio2))
```

<div style="margin-top:6px; color:#333333;">Find a lowercase i, swap it, and stop after a chosen count.</div>

<div style="background:#FFFDF8; border:2px solid #1F3A68; border-radius:14px; padding:8px; margin-top:10px;">

```python
# count char existance
iti = "information technology Institute"
print(iti)
print(iti.count("i"))
```

```python
new_iti = iti.replace("i", "$")
print(new_iti)
```

```python
iti = "information technology Institute iiiiii"
new_iti = iti.replace("i", "$")
print(new_iti)
```

```python
iti = "information technology Institute iiiiii"
new_iti = iti.replace("i", "$", 4)
print(new_iti)
```

```python
iti = "information technology Institute iiiiii"
new_iti = iti.replace("it", "$&", 4)
print(new_iti)
```

<div style="background:#FFF6D6; border-left:6px solid #C99700; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#C99700;">Look.</b> A capital <b>I</b> is not the character <b>i</b>. The third argument of <code>replace</code> is the maximum number of changes. <code>replace("it", "$&", 4)</code> only changes a real pair <code>it</code>.</div>
