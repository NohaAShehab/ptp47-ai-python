### The error is born on one line and climbs until someone catches it

<div style="margin-top:8px;">
<div style="display:inline-block; vertical-align:top; width:47%; margin-right:2%; background-color:#DFF3EA; border:2px solid #17885B; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#17885B;">Clean lines</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:10px; margin-top:8px;">"33" &nbsp;→&nbsp; int &nbsp;→&nbsp; 33<br>"324" → int → 324<br>"324" → int → 324</div>
<div style="margin-top:8px;">Every call returns. <code>load_scores</code> builds the list <b>[33, 324, 324]</b>.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:47%; background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#C9491A;">One bad line stops the list</div>
<div style="margin-top:8px; background-color:#ffffff; border-left:6px solid #B83232; border-radius:8px; padding:8px 10px; font-family:monospace;">caller<br>&nbsp;&nbsp;load_scores<br>&nbsp;&nbsp;&nbsp;&nbsp;parse_line("iti")<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;int("iti") &nbsp;✗</div>
<div style="margin-top:8px; font-size:14px;"><b>ValueError</b> is created inside <code>int</code>. <code>parse_line</code> has no <code>try</code>, so it leaves. The comprehension has no <code>try</code>, so it leaves. The cell stops on the first bad item. <code>"324"</code> is never reached.</div>
</div>
</div>

<div style="background:#FFFDF8; border:2px solid #17885B; border-radius:14px; padding:8px; margin-top:12px;">
<img src="illustrations/exc-climb.svg" alt="Clean numbers return. One bad item stops the list." style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
def parse_line(line):
    return int(line)


def load_scores(lines):
    scores = [parse_line(line) for line in lines]
    return scores

res = load_scores(["33", "324", "324"])
print(res)
```

### try / except is a net under the risky line

<div style="background-color:#E3EEFB; border:2px solid #2468B8; border-radius:12px; padding:16px; color:#111111;">
<div style="display:inline-block; vertical-align:top; width:42%; background-color:#ffffff; border-radius:8px; padding:12px; margin-right:2%;">
<div style="font-weight:700; color:#2468B8;">try</div>
<div style="font-family:monospace; margin-top:6px;">res2 = load_scores(...)</div>
<div style="font-size:14px; margin-top:6px;">This assignment finishes only if <code>load_scores</code> returns.</div>
</div>
<div style="display:inline-block; vertical-align:middle; font-size:22px; font-weight:700; color:#2468B8; margin-right:2%;">→</div>
<div style="display:inline-block; vertical-align:top; width:42%; background-color:#FCE8E0; border-radius:8px; padding:12px;">
<div style="font-weight:700; color:#C9491A;">except</div>
<div style="font-family:monospace; margin-top:6px;">print("------ Error ...")</div>
<div style="font-size:14px; margin-top:6px;">A bare <code>except:</code> catches every kind of error, including the ValueError from <code>int</code>.</div>
</div>
</div>

<div style="margin-top:12px;">
<div style="display:inline-block; vertical-align:top; width:31%; margin-right:2%; background-color:#FCE8E0; border:2px solid #C9491A; border-radius:10px; padding:12px; color:#111111;">
<div style="font-weight:700; color:#C9491A;">Name never created</div>
<div style="font-size:14px; margin-top:6px;">The error happens <b>during</b> <code>res2 = ...</code>, so the name <code>res2</code> is not assigned. The next cell's <code>print(res2)</code> is a new error: <b>NameError</b>.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:31%; margin-right:2%; background-color:#DFF3EA; border:2px solid #17885B; border-radius:10px; padding:12px; color:#111111;">
<div style="font-weight:700; color:#17885B;">Give it a fallback</div>
<div style="font-family:monospace; margin-top:6px;">except:<br>&nbsp;&nbsp;res2 = []</div>
<div style="font-size:14px; margin-top:6px;">Now the name exists either way. Printing it shows <code>[]</code>.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:31%; background-color:#E1E6EE; border:2px solid #1F3A68; border-radius:10px; padding:12px; color:#111111;">
<div style="font-weight:700; color:#1F3A68;">Or stop</div>
<div style="font-family:monospace; margin-top:6px;">except:<br>&nbsp;&nbsp;exit(2)</div>
<div style="font-size:14px; margin-top:6px;">After the message, <code>exit(2)</code> asks the program to stop instead of continuing with a bad result.</div>
</div>
</div>

<div style="background:#FFFDF8; border:2px solid #2468B8; border-radius:14px; padding:8px; margin-top:12px;">
<img src="illustrations/exc-net.svg" alt="try, except, and the cell continues" style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
def parse_line(line):
    return int(line)


def load_scores(lines):
    scores = [parse_line(line) for line in lines]
    return scores

try:
    res2 = load_scores(["gfhgfh", "324", "324"])
except :
    print('------ Error while parsing scores')


```

```python
print(res2)
```

```python

try:
    res2 = load_scores(["gfhgfh", "324", "324"])
except :
    print('------ Error while parsing scores')
    res2 = []

print(res2)
```

```python
try:
    res2 = load_scores(["gfhgfh", "324", "324"])
except :
    print('------ Error while parsing scores')
    exit(2)
```

### else is the success road. finally is the door everyone uses.

<div style="background-color:#ffffff; border:2px solid #1F3A68; border-radius:12px; padding:16px; color:#111111;">
<div style="background-color:#E3EEFB; border-radius:8px; padding:10px 12px; font-weight:700; color:#2468B8;">try &nbsp;·&nbsp; num = int(input(...))</div>
<div style="margin-top:8px;">
<div style="display:inline-block; vertical-align:top; width:48%; margin-right:2%; background-color:#DFF3EA; border-radius:8px; padding:12px;">
<div style="font-size:12px; font-weight:700; color:#17885B;">SUCCESS</div>
<div style="margin-top:6px;">no exception</div>
<div style="margin-top:8px; background-color:#ffffff; border-radius:6px; padding:8px; font-weight:700;">else &nbsp;→&nbsp; print(num)</div>
<div style="font-size:13px; margin-top:6px;">else runs only when try finished cleanly.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:48%; background-color:#FCE8E0; border-radius:8px; padding:12px;">
<div style="font-size:12px; font-weight:700; color:#C9491A;">FAILURE</div>
<div style="margin-top:6px;">ValueError, for example "wrewtet"</div>
<div style="margin-top:8px; background-color:#ffffff; border-radius:6px; padding:8px; font-weight:700;">except &nbsp;→&nbsp; print(e)</div>
<div style="font-size:13px; margin-top:6px;">You get the message. else stays quiet.</div>
</div>
</div>
<div style="margin-top:8px; background-color:#FFF6D6; border-radius:8px; padding:10px 12px; font-weight:700; color:#A85F00;">finally &nbsp;·&nbsp; always, after else or after except</div>
<div style="margin-top:8px; background-color:#E1E6EE; border-radius:8px; padding:10px 12px;">Then the lines <b>after</b> the whole try statement run, because the exception was caught.</div>
</div>

<div style="background-color:#DFF3EA; border-left:6px solid #17885B; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#17885B;">Order.</b> try, then except <b>or</b> else, then finally, then the rest of the cell. finally is for closing a file or showing "we reached the end", not for the main result.</div>

<div style="background:#FFFDF8; border:2px solid #A85F00; border-radius:14px; padding:8px; margin-top:12px;">
<img src="illustrations/exc-roads.svg" alt="else, except, and finally" style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
try:
    num = int(input("Enter a number: "))
except Exception as e:
    # this code will be executed if there are errors in any line of try block
    print(e)
else:
    # this code will be executed only if the process executed successfully
    print(num)

```

### finally runs on the way out, even when the function returns

<div style="background-color:#E1E6EE; border:2px solid #1F3A68; border-radius:12px; padding:16px; color:#111111;">
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:12px;">
try: &nbsp;&nbsp;&nbsp;&nbsp;read two numbers, return their sum<br>
except: &nbsp;print the error, return None<br>
finally: print "reached its end"<br>
<span style="color:#B83232;">print("***") &nbsp;&nbsp;← never runs</span>
</div>
<div style="margin-top:10px; font-size:14px;">Both roads <b>return</b>, so the function leaves. The stars sit after that, with no path leading to them. finally still prints, because Python runs it before the return actually leaves.</div>
<div style="margin-top:12px;">
<div style="display:inline-block; vertical-align:top; width:47%; margin-right:2%; background-color:#DFF3EA; border-radius:8px; padding:12px;">
<div style="font-weight:700; color:#17885B;">Bad input "wqewr"</div>
<div style="margin-top:6px;">except returns <b>None</b></div>
<div style="margin-top:4px;">finally prints, then the caller sees <b>None</b></div>
</div>
<div style="display:inline-block; vertical-align:top; width:47%; background-color:#DFF3EA; border-radius:8px; padding:12px;">
<div style="font-weight:700; color:#17885B;">Good input, sum 478</div>
<div style="margin-top:6px;">try returns <b>478</b></div>
<div style="margin-top:4px;">finally prints, then the caller sees <b>478</b></div>
</div>
</div>
</div>

<div style="background-color:#FCE8E0; border:2px solid #B83232; border-radius:12px; padding:16px; margin-top:12px; color:#111111;">
<div style="font-weight:700; color:#B83232;">A return inside finally replaces the other return</div>
<div style="margin-top:10px; font-family:monospace; background-color:#ffffff; border-radius:8px; padding:12px;">
try returns &nbsp;&nbsp;&nbsp;6110<br>
finally returns 1000 &nbsp;&nbsp;<span style="color:#B83232;">← this one wins</span><br>
caller receives 1000
</div>
<div style="margin-top:8px; font-size:14px;">Python warns: <b>SyntaxWarning: 'return' in a 'finally' block</b>. The sum was computed and then thrown away. Keep results in try or except. Use finally for cleanup, not for a second return.</div>
</div>


```python

def sum_num():
    try:
        num1 = int(input("Enter a number: "))
        num2 = int(input("Enter another number: "))
        print(num1 + num2)
        res = num1 + num2
        return res
    except Exception as e:
        print(e)
        return None
    finally:
        print("===The function reached its end  .... ")
    print("*******************************************")

val = sum_num()

print(val)
```

### raise throws an exception yourself. Some operations throw their own.

<div style="margin-top:8px;">
<div style="display:inline-block; vertical-align:top; width:47%; margin-right:2%; background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#C9491A;">You raise it</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:10px; margin-top:8px;">if num2 == "0":<br>&nbsp;&nbsp;raise ZeroDivisionError(<br>&nbsp;&nbsp;&nbsp;&nbsp;"You entered num2 = 0....")</div>
<div style="margin-top:8px; font-size:14px;">The division line is never reached. The message is the one you wrote. The caller sees a traceback that points at <code>raise</code>.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:47%; background-color:#E1E6EE; border:2px solid #1F3A68; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#1F3A68;">Python raises it</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:10px; margin-top:8px;">res = int(num1) / int(num2)</div>
<div style="margin-top:8px; font-size:14px;">Take the <code>raise</code> away and type <code>0</code> for the second number. Both strings pass <code>isdigit()</code>, then <code>/</code> itself raises <b>ZeroDivisionError: division by zero</b>.</div>
</div>
</div>

<div style="background-color:#E3EEFB; border-left:6px solid #2468B8; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#2468B8;">Same type, two births.</b> <code>raise</code> is how a function reports a problem it detected. The <code>/</code> operator reports a problem it detected. Either way the value that climbs is a <code>ZeroDivisionError</code>, and a matching <code>except</code> could catch it.</div>

<div style="background-color:#0F1724; border-radius:12px; padding:16px 18px; margin-top:14px;">
<div style="color:#FFD43B; font-weight:700;">Picture to keep</div>
<div style="color:#F6F4EE; margin-top:8px; line-height:1.55;">An exception climbs until an <span style="color:#FFD43B;">except</span> of the right type accepts it. <span style="color:#9FD9B8;">else</span> runs only on success. <span style="color:#FFD43B;">finally</span> runs on the way out, and its return would replace the one you wanted. <span style="color:#F6B59A;">raise</span> is you sending that climb on purpose.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #C9491A; border-radius:14px; padding:8px; margin-top:12px;">
<img src="illustrations/exc-raise.svg" alt="raise, ZeroDivision, and except" style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
def div_num():
    num1 = input("Enter a number: ")
    num2 = input("Enter another number: ")
    if num2 =='0': raise ZeroDivisionError(f"You entered num2 = {num2}....")
    if num1.isdigit() and num2.isdigit():
        res = int(num1) / int(num2)
        return res
    return None

resss =div_num()









```
