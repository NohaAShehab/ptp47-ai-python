```python
class Course:
    all_courses = {}
    def __init__(self, name, price):
        self.name = name
        self.price = price
        self.__class__.all_courses[name] = self
    # we need to say that this function doesn't depend on the instance , its requires class itself
    def search_by_name(self, name ):
        return Course.all_courses.get(name) | None


c1 = Course('python', 1000)
c2 = Course('data', 303003)
print(Course.all_courses)
```

```python
class Course:
    all_courses = {}
    def __init__(self, name, price):
        self.name = name
        self.price = price
        self.__class__.all_courses[name] = self
    # we need to say that this function doesn't depend on the instance , its requires class itself
    @classmethod # Usually first parameter of such methods is named cls
    def search_by_name(self, name ):
        print(self)
        return Course.all_courses.get(name)


res = Course.search_by_name("python")
print(res)
```

```python
class Course:
    all_courses = {}
    def __init__(self, name, price):
        self.name = name
        self.price = price
        self.__class__.all_courses[name] = self
    # we need to say that this function doesn't depend on the instance , its requires class itself
    @classmethod # Usually first parameter of such methods is named cls
    def search_by_name(cls, name ):
        print(cls)
        # return Course.all_courses.get(name)
        return cls.all_courses.get(name)


res = Course.search_by_name("python")
print(res)
c = Course('python', 1000)
res = Course.search_by_name("python")
print(res)
```

You did not have to remember the host and port.</div></div>

```python
class DB_Connection:
    def __init__(self, db_name, port):
        self.db_name = db_name
        self.port = port

    @classmethod
    def default_connection(cls):
        return cls("postgres",345)


conn = DB_Connection("iti",23)
print(isinstance(conn, DB_Connection))

conn1 = DB_Connection.default_connection()
print(isinstance(conn1, DB_Connection))



```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def cal_net_salary(salary):
        return salary*.8


obj = Employee("Noha", 5000)
obj.cal_net_salary(54545)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def cal_net_salary(salary):
        print(salary)
        # return salary*.8


obj = Employee("Noha", 5000)
obj.cal_net_salary()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    @staticmethod  # this function doesn't depend on instancce or class --> it is just a helper
    def cal_net_salary(salary):
        print(salary)
        # return salary*.8


obj = Employee("Noha", 5000)
obj.cal_net_salary(obj.salary)
Employee.cal_net_salary(2124235)
```
