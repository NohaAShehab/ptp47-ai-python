```python


####
# function accept argument, return res

def sum_num(num1, num2):
    print(f"num1= {num1}, num2= {num2}")
    res = num1 + num2
    return res

myres = sum_num(3,4)
print(myres)

myres2 = sum_num(4,3)
print(myres2)



myres2 = sum_num(num2=4,num1=3)
print(myres2)


```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> Each call prints the two numbers, then <code>7</code>. In the last call the names choose the parameters, so writing num2 first still stores 3 in num1.</div>
