```python
class Student:
    def __init__(self, name ,grade):
        self.name = name
        self.grade = grade

    def __str__(self): # must return string --> with end user
        return f"{self.name}"
    #
    def __repr__(self) -> str:
        return f"Studnet(name={self.name}, grade={self.grade})"



s = Student("Hoda", 100)
print(s)
print(s.__repr__())
print(id(s))
```

<img src="illustrations/tips-eq.jpg" alt="Two separate Hoda cards are not equal until same name and same grade make them match" style="width:100%; border-radius:16px; margin-top:14px;">

```python
class Student:
    def __init__(self, name ,grade):
        self.name = name
        self.grade = grade


s = Student("Hoda", 100)
s2 = Student("Hoda", 100)
print(s == s2 ) # compare address of s and s2

s3 = s
print(s3 == s )
```

```python
#
class Student:
    def __init__(self, name ,grade):
        self.name = name
        self.grade = grade

    def __eq__(self, other):
        if self.grade == other.grade and self.name == other.name:
            return True
        return False



s = Student("Hoda", 100)
s2 = Student("Hoda", 100)
print(s == s2 ) # compare address of s and s2

s3 = s
print(s3 == s )
```

<div style="background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:14px 18px; color:#111111;"><span style="display:inline-block; background-color:#C9491A; color:#ffffff; font-weight:700; border-radius:16px; width:32px; line-height:32px; text-align:center; margin-right:10px;">4</span><span style="font-size:20px; font-weight:700;">Delete the object</span><div style="margin-top:8px; font-size:15px; line-height:1.45;">ارقد بسلام يا أبو الأحناف<br><br>Four fish have been counted. <code>del f3</code> drops a name. Without <code>__del__</code>, the class counter never hears about it and stays at 4. With <code>__del__</code>, that fish says goodbye and the count falls to 3.</div></div>

<img src="illustrations/tips-del.jpg" alt="Four fish counted as 4, with the green fish fading under del and the count moving to 3" style="width:100%; border-radius:16px; margin-top:14px;">

```python
class Fish:
    count = 0
    def __init__(self, color):
        self.color = color
        self.__class__.count += 1


f1 = Fish("red")
f2 = Fish("blue")
print(
    Fish.count
)
f3 = Fish("green")
f4 = Fish("purple")
print( Fish.count )
```

```python
del f3
```

```python
print( Fish.count )
```

```python
class Fish:
    count = 0
    def __init__(self, color):
        self.color = color
        self.__class__.count += 1

    def __del__(self):
        self.__class__.count -= 1

f1 = Fish("red")
f2 = Fish("blue")
print(
    Fish.count
)
f3 = Fish("green")
f4 = Fish("purple")
print( Fish.count )
del f3
print( Fish.count )
```

```python
class Fish:
    count = 0
    def __init__(self, color):
        self.color = color
        self.__class__.count += 1

    def __del__(self):
        self.__class__.count -= 1

    def __len__(self):
        """
            then len(obj)
        :return: the number of properties in the object
        """
        # must return with int
        return len(self.__dict__)


f1 = Fish("red")
f1.loc = 'alex'
print(len(f1))
```

<div style="background-color:#FFF6D6; border:2px solid #A85F00; border-radius:12px; padding:14px 18px; color:#111111;"><span style="display:inline-block; background-color:#A85F00; color:#ffffff; font-weight:700; border-radius:16px; width:32px; line-height:32px; text-align:center; margin-right:10px;">6</span><span style="font-size:20px; font-weight:700;">Appending the list does not open a slot</span><div style="margin-top:8px; font-size:15px; line-height:1.45;">The layout is fixed when the class is created. Appending z onto <code>Point.__slots__</code> changes the list you can see, before or after <code>Point(2, 3)</code>, and z is still refused. A tuple of slots cannot even be appended.</div></div>

```python
class Point:
    __slots__ = ['x', 'y']
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(2,3)
Point.__slots__.append('z')
print(Point.__dict__)
p.z = 'werew'
```

```python
class Point:
    __slots__ = ['x', 'y']
    def __init__(self, x, y):
        self.x = x
        self.y = y
Point.__slots__.append('z')
p = Point(2,3)
print(Point.__dict__)
p.z = 'werew'
```
