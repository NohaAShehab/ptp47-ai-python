<span style="font-size:22px; font-weight:700;">Build the record, then change it</span>
<div style="margin-top:6px; color:#333333;">len counts keys. A new name adds a key. An old name replaces the value.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #2468B8; border-radius:14px; padding:8px; margin-top:10px;">
<img src="illustrations/dict-edit.svg" alt="name changed to Linda Deif and salary added" style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
info = {
    "name" : "Linda",
    "age" : 22,
    "city" : "Ras-Elbar",
    "track" : "AI"
}
```

```python
print(len(info))
```

```python
info["salary"] = 100000

print(info)
```

```python
info['name'] = 'Linda Deif'

print(info)
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>len(info)</code> is 4. After salary is added it would be 5. <code>info['name'] = 'Linda Deif'</code> replaces Linda. The other keys stay.</div>


<div id="s2" style="background:#DFF3EA; border:2px solid #17885B; border-radius:14px; padding:14px 18px; color:#111111;">
<span style="display:inline-block; background:#17885B; color:#ffffff; font-weight:700; border-radius:16px; min-width:34px; line-height:34px; text-align:center; margin-right:10px;">2</span>
<span style="font-size:22px; font-weight:700;">Loop, keys, and values</span>
<div style="margin-top:6px; color:#333333;">A for-loop walks the keys. The view looks like a list and is not one.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #17885B; border-radius:14px; padding:8px; margin-top:10px;">
<img src="illustrations/dict-views.svg" alt="dict_keys view and a TypeError on keys[0]" style="width:100%; height:auto; display:block; border-radius:10px;">
</div>


```python
for k in info:
    print(f"Key = {k} Value = {info[k]}")
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> The loop prints name, age, city, track, and salary, each with its value.</div>


<div style="background:#FDECEC; border-left:6px solid #B83232; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#B83232;">This cell is meant to fail.</b> <code>keys[0]</code> raises <code>TypeError</code>: 'dict_keys' object is not subscriptable. <code>keys</code> is already stored before that line, so later cells can still loop over it. <code>values = info.values()</code> does not run.</div>


```python
keys=  info.keys()
print(keys)
print(keys[0])
values = info.values()
```

```python
for v in info.values():
    print(v)
```

```python
for k in keys:
    print(k)
```

<div style="background:#FDECEC; border-left:6px solid #B83232; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#B83232;">This cell is meant to fail.</b> Same rule again. <code>keys[0]</code> is still not a list index.</div>


```python
print(keys[0])
```

<div id="s3" style="background:#FCE8E0; border:2px solid #C9491A; border-radius:14px; padding:14px 18px; color:#111111;">
<span style="display:inline-block; background:#C9491A; color:#ffffff; font-weight:700; border-radius:16px; min-width:34px; line-height:34px; text-align:center; margin-right:10px;">3</span>
<span style="font-size:22px; font-weight:700;">items gives both pieces</span>
<div style="margin-top:6px; color:#333333;">Each step of the loop receives the key and the value.</div>
</div>


```python
items =  info.items()
print(items)
```

```python
items_list = list(items)
```

```python
for k, v in items:
    print(f"key = {k} value = {v}")
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>info.items()</code> is a view of pairs. <code>list(items)</code> makes a real list of those pairs. The loop prints key and value on one line.</div>


<div id="s4" style="background:#E1E6EE; border:2px solid #1F3A68; border-radius:14px; padding:14px 18px; color:#111111;">
<span style="display:inline-block; background:#1F3A68; color:#ffffff; font-weight:700; border-radius:16px; min-width:34px; line-height:34px; text-align:center; margin-right:10px;">4</span>
<span style="font-size:22px; font-weight:700;">Build a dict from pairs</span>
<div style="margin-top:6px; color:#333333;">dict() reads a sequence of sequences. Each inner one needs two items.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #1F3A68; border-radius:14px; padding:8px; margin-top:10px;">

```python
myd = dict([
    ('name', 'Sofyan'), ("track", "ai")
])
print(myd)






```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> This one works: {'name': 'Sofyan', 'track': 'ai'}.</div>


<div style="background:#FDECEC; border-left:6px solid #B83232; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#B83232;">This cell is meant to fail.</b> <code>('track', 'ai', 47)</code> has length 3. <code>ValueError</code>: dictionary update sequence element #1 has length 3; 2 is required.</div>


```python
myd = dict([
    ('name', 'Sofyan'), ("track", "ai", 47)
])
print(myd)
```

<div style="background:#FDECEC; border-left:6px solid #B83232; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#B83232;">This cell is meant to fail.</b> <code>(47,)</code> has length 1. <code>ValueError</code>: dictionary update sequence element #2 has length 1; 2 is required.</div>


```python
myd = dict([
    ('name', 'Sofyan'), ("track", "ai"),(47,)
])
print(myd)
```

<span style="display:inline-block; background:#6B4C9A; color:#ffffff; font-weight:700; border-radius:16px; min-width:34px; line-height:34px; text-align:center; margin-right:10px;">5</span>
<span style="font-size:22px; font-weight:700;">update merges another mapping</span>
<div style="margin-top:6px; color:#333333;">Shared keys are overwritten. New keys are added.</div>

<div style="background:#FFFDF8; border:2px solid #6B4C9A; border-radius:14px; padding:8px; margin-top:10px;">
<img src="illustrations/dict-update.svg" alt="age 22 replaced by age 17 when courses is merged" style="width:100%; height:auto; display:block; border-radius:10px;">

```python
courses = {
    "foundation" : ["python", "DB"],
    "Core": ["Ml", "DL"],
    "age": 17
}
info.update(courses)
print(info)


```
