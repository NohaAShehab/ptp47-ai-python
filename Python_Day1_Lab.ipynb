| Step | What you write |
| --- | --- |
| Understand | In: one whole number of seconds. Out: a line such as `2 min 15 s`. |
| By hand | 135 seconds: 135 // 60 is 2 minutes, and 135 % 60 leaves 15 seconds. |
| Plan | Read the total. Minutes = total // 60. Seconds = total % 60. Print both. |
| Test | 135 gives `2 min 15 s`, 59 gives `0 min 59 s`, 60 gives `1 min 0 s`. |

Run the cell below and try the three test values.

````python
total = int(input("Seconds: "))
minutes = total // 60
seconds = total % 60
print(f"{minutes} min {seconds} s")
````
