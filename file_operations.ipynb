```python
# write arabic content

with open("story.txt", "w") as file_object:
    file_object.write("صباح الخير يا اى تى اى .. \nانا جاى غصب عنى سيبونى أنام يومين وبعدين هنخلص الامتحانات والتاسك وكل حاجة")


with open("story.txt", "r") as file_object:
    data = file_object.read()
    print(data)
```

```python
info = {
    'first_name' :  'Menna',
    'track' : 'AI',
    'salary' : 200000 ,
    'currency' : 'KWD'
}
```

```python
with open("students.txt", "a") as file_object:
    file_object.write(info)
```

<span style="font-size:20px; font-weight:700;">JSON keeps the shape</span>
<div style="margin-top:8px;"><code>json.dumps</code> and <code>json.dump</code> turn the dictionary into text. <code>json.load</code> turns that text back into keys and values.</div>
</div>

<div style="background:#FFFDF8; border:2px solid #17885B; border-radius:14px; padding:8px; margin-top:12px;">
<img src="illustrations/files-json.svg" alt="A dictionary becomes JSON text and is loaded back into a dictionary" style="width:100%; height:auto; display:block; border-radius:10px;">

```python
# json demo
import  json

info = {
    'first_name' :  'Menna',
    'track' : 'AI',
    'salary' : 200000 ,
    'currency' : 'KWD'
}

data = json.dumps(info)
print(data, type(data))
```

```python
with open("students.json", "a") as file_object:
    json.dump(info, file_object)
```

```python
def read_data_from_json(file_name : str):
    data = []
    try:
        with open(file_name, "r") as file_object:
            data = json.load(file_object) # convert json string to list of dict

    except Exception as e:
        print(e)

    return data


students_data = read_data_from_json("students.json")
print(students_data)


```

```python
def save_data_to_json(file_name :str, user_object: dict ):
    old_data = read_data_from_json(file_name)
    print(old_data)
    old_data.append(user_object)
    print(old_data)

    try:
        with open(file_name, "w") as file_object:
            json.dump(old_data, file_object, indent=4)
            return True
    except Exception as e:
        print(e)
        return False


ali_info = {
    "first_name" : "Ali",
    "track" : "AI",
    "salary" : 1000000 ,
    "tax" : None
}

save_info = save_data_to_json("students.json", ali_info)






```
