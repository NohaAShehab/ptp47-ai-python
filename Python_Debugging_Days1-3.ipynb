## The errors you will meet

| Error | Usually means |
| --- | --- |
| `SyntaxError` | Python cannot read the code: missing colon, bracket or quote |
| `IndentationError` | Lines of one block are not lined up |
| `NameError` | A name that was never defined, often a typo |
| `TypeError` | The operation does not work for this type, such as "5" + 5 |
| `ValueError` | Right type, wrong value, such as int("abc") |
| `IndexError / KeyError` | A position or key that does not exist |
| `AttributeError` | The object has no such method or attribute |
| `UnboundLocalError` | A name used before it is assigned in the same function |
| `ZeroDivisionError` | Division or % by zero |
| `ModuleNotFoundError / ImportError` | The module, or the name inside it, cannot be found |

Print the odd numbers from 1 to 9. The loop has a safety stop so it cannot freeze the notebook; without it, it would run forever.

**Expected output after your fix**

```text
1 3 5 7 9 
```

<div style="background-color:#FDE7E7; border-left:6px solid #B83232; border-radius:6px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#B83232;">Watch out:</b> if a cell never finishes, press <b>Stop</b> (interrupt the kernel). Real infinite loops have no safety stop.</div>

````python
i = 0
steps = 0
while i < 10:
    steps += 1
    if steps > 30:
        raise RuntimeError("stopped after 30 steps: the loop never ends")
    if i % 2 == 0:
        continue
    print(i, end=" ")
    i += 1
````

<span style="display:inline-block; background-color:#BFE8D3; color:#111111; border-radius:4px; padding:2px 10px; font-weight:700; font-size:13px; letter-spacing:1px;">MY ANSWER · A8</span> &nbsp; double-click to edit

When does each times function read n: when it is defined, or when it is called?

*Your answer:* 

````python
def make_multipliers():
    funcs = []
    for n in range(1, 4):
        def times(x):
            return x * n
        funcs.append(times)
    return funcs

for f in make_multipliers():
    print(f(10))
````

<div style="background-color:#E1E6EE; border-left:6px solid #1F3A68; border-radius:6px; padding:8px 12px; color:#111111;"><span style="display:inline-block; background-color:#1F3A68; color:#ffffff; border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">D2</span><span style="display:inline-block; background-color:#A85F00; color:#ffffff; border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">Wrong output, no error</span><span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #1F3A68;border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">initial values</span><span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #1F3A68;border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">edge cases</span><span style="display:inline-block; background-color:#A85F00; color:#ffffff; border-radius:12px; padding:1px 10px; margin-right:6px; font-size:13px; font-weight:600;">Level: Medium</span></div>

`extremes(values)` should return the smallest and the largest value. It never crashes, but the test cell finds wrong answers.

**Expected output after your fix**

```text
(7, 30)
(-9, -2)
```

````python
def extremes(values):
    smallest = 0
    biggest = 0
    for v in values:
        if v < smallest:
            smallest = v
        if v > biggest:
            biggest = v
    return smallest, biggest

print(extremes([12, 7, 30]))
print(extremes([-5, -2, -9]))
````

````python
students = [("Mariam", "91"), ("Omar", "78"), ("Salma", "abc"), ("Nour", "88.5")]
passed = []

def report(records):
    marks = []
    for name, text in records:
        try:
            mark = int(text)
        except ValueError:
            print("skipping", name)
            continue
        marks.append(mark)
        if mark >= 80:
            passed.append(name)
    print("Average:", round(sum(marks) / len(records), 1))
    print("Passed:", ", ".join(passed))

report(students)
report(students)
````
