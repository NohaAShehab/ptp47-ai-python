### A function inside a function

<div style="background-color:#E3EEFB; border:2px solid #2468B8; border-radius:12px; padding:16px 18px; color:#111111;">
<div style="font-size:13px; font-weight:700; letter-spacing:1px; color:#2468B8;">ENCLOSING SCOPE · make_scaler(44, 5)</div>
<div style="margin-top:10px; background-color:#ffffff; border-radius:8px; padding:10px 12px;">
<div><span style="display:inline-block; background-color:#2468B8; color:#ffffff; border-radius:6px; padding:2px 8px; font-weight:700;">mean</span> &nbsp; 44</div>
<div style="margin-top:6px;"><span style="display:inline-block; background-color:#2468B8; color:#ffffff; border-radius:6px; padding:2px 8px; font-weight:700;">std</span> &nbsp; 5</div>
<div style="margin-top:12px; background-color:#DFF3EA; border:2px dashed #17885B; border-radius:10px; padding:12px;">
<div style="font-size:13px; font-weight:700; color:#17885B;">LOCAL SCOPE · scaler(x=10)</div>
<div style="margin-top:6px; font-family:monospace;">(x - mean) / std</div>
<div style="margin-top:6px; font-family:monospace;">(10 - 44) / 5 &nbsp;=&nbsp; <b>-6.8</b></div>
</div>
</div>
</div>

<div style="background-color:#FFF6D6; border-left:6px solid #C99700; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#A85F00;">What the next cell does.</b> <code>scaler()</code> is called <b>before</b> <code>make_scaler</code> finishes, so you see the number printed from the inside. Nothing is returned. The parentheses are the difference between running a function and holding onto it.</div>


```python
def make_scaler(mean, std):
    def scaler(x=10):
        res =  (x - mean) / std
        print(res)

    scaler()

dd = make_scaler(44, 5)
```

### Hand back a number, or hand back the function

<div style="margin-top:8px;">
<div style="display:inline-block; vertical-align:top; width:47%; margin-right:2%; background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#C9491A;">return scaler()</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:8px 10px; margin-top:8px;">call now → return the number</div>
<div style="margin-top:10px; font-size:14px;">The parentheses run <code>scaler</code> at once, with <code>x=10</code>.</div>
<div style="margin-top:8px; background-color:#ffffff; border-radius:8px; padding:8px 10px; font-family:monospace;">res = -0.06126…<br>type: float</div>
<div style="margin-top:8px; font-size:14px;">You can print <code>res</code>. You cannot call it again. The tool was used up inside <code>make_scaler</code>.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:47%; background-color:#DFF3EA; border:2px solid #17885B; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#17885B;">return scaler</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:8px 10px; margin-top:8px;">return the function → call it later</div>
<div style="margin-top:10px; font-size:14px;">No parentheses. <code>res</code> <b>is</b> <code>scaler</code>, still tied to <code>mean</code> and <code>std</code>.</div>
<div style="margin-top:8px; background-color:#ffffff; border-radius:8px; padding:8px 10px; font-family:monospace;">res&nbsp;&nbsp;→ function<br>res() → -0.06126…</div>
<div style="margin-top:8px; font-size:14px;">A second factory, <code>make_scaler(100, 55)</code>, remembers a different pair. <code>new_scale(5)</code> uses those remembered values with a new <code>x</code>.</div>
</div>
</div>

<div style="background-color:#E3EEFB; border-left:6px solid #2468B8; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;">
<b style="color:#2468B8;">Same recipe, two results.</b>
<span style="display:inline-block; background-color:#2468B8; color:#ffffff; border-radius:12px; padding:1px 10px; margin:4px 4px 0 6px; font-size:13px;">make_scaler(44, 555)</span>
<span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #2468B8; border-radius:12px; padding:1px 10px; margin:4px 4px 0 0; font-size:13px;">x = 10</span>
<span style="display:inline-block; background-color:#ffffff; color:#111111; border:1px solid #2468B8; border-radius:12px; padding:1px 10px; margin:4px 4px 0 0; font-size:13px;">(10 − 44) / 555</span>
</div>


```python
def make_scaler(mean, std):
    def scaler(x=10):
        res = (x - mean) / std
        return res

    return scaler()  # when return with -->call scaler

res = make_scaler(44,555)
```

```python
print(res)
```

### nonlocal keeps the enclosing names writable

<div style="background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:16px; color:#111111;">
<div style="font-size:13px; font-weight:700; letter-spacing:1px; color:#C9491A;">counter() BUILDS THIS, THEN RETURNS step</div>
<div style="margin-top:12px;">
<div style="display:inline-block; vertical-align:top; width:30%; margin-right:2%; background-color:#ffffff; border-top:6px solid #2468B8; border-radius:8px; padding:10px 12px;">
<div style="font-size:12px; color:#2468B8; font-weight:700;">CELL · count</div>
<div style="font-family:monospace; font-size:18px; margin-top:4px;">0 → 1 → 2</div>
<div style="font-size:13px; margin-top:4px;"><code>step</code> does <code>count += 1</code></div>
</div>
<div style="display:inline-block; vertical-align:top; width:30%; margin-right:2%; background-color:#ffffff; border-top:6px solid #17885B; border-radius:8px; padding:10px 12px;">
<div style="font-size:12px; color:#17885B; font-weight:700;">CELL · data</div>
<div style="font-family:monospace; font-size:18px; margin-top:4px;">"test"</div>
<div style="font-size:13px; margin-top:4px;">read, not changed</div>
</div>
<div style="display:inline-block; vertical-align:top; width:30%; background-color:#ffffff; border-top:6px solid #1F3A68; border-radius:8px; padding:10px 12px;">
<div style="font-size:12px; color:#1F3A68; font-weight:700;">CELL · unit</div>
<div style="font-family:monospace; font-size:18px; margin-top:4px;">5000</div>
<div style="font-size:13px; margin-top:4px;">read, not changed</div>
</div>
</div>
<div style="margin-top:12px; background-color:#ffffff; border-radius:8px; padding:10px 12px; font-family:monospace;">a = counter() &nbsp;→&nbsp; a is step<br>a() &nbsp;→&nbsp; 1<br>a() &nbsp;→&nbsp; 2</div>
</div>

<div style="background-color:#FFF6D6; border-left:6px solid #C99700; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#A85F00;">Why nonlocal.</b> <code>count += 1</code> assigns to <code>count</code>. Assignment would make <code>count</code> local, and the local name would not exist yet. <code>nonlocal</code> says: assign to the enclosing cell, do not create a new local.</div>

<div style="background-color:#E1E6EE; border-left:6px solid #1F3A68; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#1F3A68;">Reading __closure__.</b> Each cell is one remembered name. Python stores them as <code>count</code>, <code>data</code>, <code>unit</code>, so index 0 is the integer <code>count</code> (2 after two calls), index 1 is <code>"test"</code>, index 2 is <code>5000</code>.</div>


```python
def counter():
    data = 'test'

    count =0
    unit  = 5000
    def step():
        nonlocal data
        nonlocal count
        nonlocal unit

        count += 1
        return count

    return step

a = counter()
print(a.__closure__)
print(a())
a()
print(a.__closure__[0].cell_contents)
print(a.__closure__[1].cell_contents)
print(a.__closure__[2].cell_contents)






```

### A type hint is a label. The operator is the guard.

<div style="margin-top:8px;">
<div style="display:inline-block; vertical-align:top; width:47%; margin-right:2%; background-color:#E3EEFB; border:2px solid #2468B8; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#2468B8;">What you see in the header</div>
<div style="font-family:monospace; background-color:#ffffff; border-radius:8px; padding:10px; margin-top:8px;">def sum_num(num1: int, num2: int) -&gt; int</div>
<div style="margin-top:8px; font-size:14px;">The <code>: int</code> and <code>-&gt; int</code> are notes for you and for tools. Python does not check them when the function runs.</div>
</div>
<div style="display:inline-block; vertical-align:top; width:47%; background-color:#FCE8E0; border:2px solid #C9491A; border-radius:12px; padding:14px 16px; color:#111111;">
<div style="font-weight:700; color:#C9491A;">What + actually does</div>
<div style="margin-top:8px; background-color:#ffffff; border-radius:8px; padding:8px 10px;"><span style="color:#17885B; font-weight:700;">int + int</span> &nbsp; 10 + 20 &nbsp;→&nbsp; 30</div>
<div style="margin-top:6px; background-color:#ffffff; border-radius:8px; padding:8px 10px;"><span style="color:#A85F00; font-weight:700;">str + str</span> &nbsp; "10" + "iti" &nbsp;→&nbsp; "10iti"</div>
<div style="margin-top:6px; background-color:#ffffff; border-radius:8px; padding:8px 10px;"><span style="color:#B83232; font-weight:700;">int + str</span> &nbsp; 10 + "20" &nbsp;→&nbsp; TypeError</div>
</div>
</div>

<div style="background-color:#FFF6D6; border-left:6px solid #C99700; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#A85F00;">Same story twice.</b> The next cells run <code>sum_num</code> with no hints, then again with hints. The three results do not change. The hint did not stop <code>"10" + "iti"</code>, and it did not stop the <code>TypeError</code>. The error comes from <code>+</code>, not from the annotation.</div>


```python
def sum_num(num1, num2):
    return num1 + num2


print(sum_num(10,20))
```

```python
print(sum_num("10","iti"))
```

```python
print(sum_num(10,"20"))
```

```python
def sum_num(num1 :int, num2 :int) ->int:
    return num1 + num2


print(sum_num(10,20))



```

```python
print(sum_num("10","iti"))
```

### isinstance is a check that actually runs

<div style="background-color:#DFF3EA; border:2px solid #17885B; border-radius:12px; padding:16px; color:#111111;">
<div style="font-family:monospace; font-size:15px;">sum_num(num1, num2)</div>
<div style="margin-top:12px; background-color:#ffffff; border-radius:8px; padding:12px;">
<div style="font-weight:700;">Are both values ints?</div>
<div style="margin-top:10px;">
<div style="display:inline-block; vertical-align:top; width:46%; margin-right:3%; background-color:#DFF3EA; border-radius:8px; padding:10px 12px;">
<div style="font-weight:700; color:#17885B;">yes</div>
<div style="font-family:monospace; margin-top:6px;">return num1 + num2</div>
<div style="margin-top:4px; font-size:14px;">10 and 20 → 30</div>
</div>
<div style="display:inline-block; vertical-align:top; width:46%; background-color:#FCE8E0; border-radius:8px; padding:10px 12px;">
<div style="font-weight:700; color:#C9491A;">no</div>
<div style="font-family:monospace; margin-top:6px;">return None</div>
<div style="margin-top:4px; font-size:14px;">the function refuses the mix instead of crashing</div>
</div>
</div>
</div>
</div>

<div style="background-color:#E3EEFB; border-left:6px solid #2468B8; border-radius:6px; padding:10px 14px; margin-top:10px; color:#111111;"><b style="color:#2468B8;">The header matches the body.</b> <code>-&gt; int | None</code> describes the two roads. The <code>isinstance</code> test is what chooses the road. The docstring says the same thing in words.</div>


```python
def sum_num(num1 :int, num2 :int) ->int|None :
    """
    :param num1: first number
    :param num2: second number
    :return: the summation of num1 and num2 or none
    """
    if isinstance(num1, int) and isinstance(num2, int):
        return num1 + num2
    return None


print(sum_num(10,20))











```
