It previews copying: a shallow copy shares inner lists, and <code>[[0] * 3] * 2</code> repeats one inner list. The drawing and the live grid come in the next section.</div>


```python
# copy() and slicing make a new outer list, but its slots still point to the same inner lists, so changing shallow[0][0] changes grid too. copy.deepcopy() copies every level. The grid trap is common in matrix and game exercises: [[0] * 3] * 2 repeats a reference to one inner list, so both rows change together. The fix is a comprehension, board = [[0] * 3 for _ in range(2)], which builds a separate inner list for each row. For numerical grids students will later use NumPy arrays instead, which do not have this problem.
```

```python
courses = ["python", "ML", "Database", "Generative"]
```

```python
courses.sort(reverse=True)
print(courses)
```

```python
courses.sort(key=str.lower)
```

```python
print(courses)
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>sort()</code> puts Database, Generative, ML, python. <code>reverse=True</code> flips that order. <code>key=str.lower</code> ignores case and returns to Database, Generative, ML, python.</div>

The grid trap is common in matrix and game exercises: <code>[[0] * 3] * 2</code> repeats a reference to one inner list, so both rows change together. The fix is a comprehension, <code>board = [[0] * 3 for _ in range(2)]</code>, which builds a separate inner list for each row. For numerical grids, NumPy arrays do not have this problem.</div>


```python
l1 = ["AI", "ITI", 47, 'Mansoura']
print(l1)
print(id(l1))
```

```python
l2 = l1.copy()
print(l2)
print(id(l2))
```

```python
# lecture example
grid = [[0,0], [0,0]]
print(id(grid))

shallow = grid.copy()
print(id(shallow))

```

```python
grid.append("iti")
```

```python
grid[0][0]= 10
```

```python
print(grid)
print(shallow)
```

```python
shallow.append("hello")
print(shallow)
print(grid)
```

```python
shallow[1][1]= 100

print(shallow)
print(grid)
```

```python
shallow.append(["iti", "rer"])
print(shallow)
print(grid)
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> Appending <code>iti</code> or <code>hello</code> changes only that outer list. <code>grid[0][0] = 10</code> and <code>shallow[1][1] = 100</code> show up in both names, because the inner lists are shared. The ids of the two outer lists differ.</div>

```python
l1 = [342,24,2323]
l2 = ["erret", "ewrwerrw"]
l3 = l1 + l2  # return new list
print(l3 )







```

```python
l1.extend(l2)
print(l2)
```

```python
print(l1)
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>l3</code> is [342, 24, 2323, 'erret', 'ewrwerrw']. <code>extend</code> writes those same items onto <code>l1</code> and leaves <code>l2</code> as it was. The cell prints <code>l2</code>, so the screen still shows the short list.</div>

```python
# print item in courses and its index ?

index = 0
for course in courses:
    print(course, index)
    index +=1






```

```python
course_enum = enumerate(courses)
print(course_enum)
```

```python
# iterabole object (index, val)
for i , v in course_enum:
    print(i, v)
```

```python
for i , v in course_enum:
    print(i, v)
```
