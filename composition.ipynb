```python
from types import TracebackType


class Engine:
    def __init__(self, hp):
        self.hp  = hp

    def start(self):
        return f"Engine start with {self.hp}"

class Car:
    def __init__(self, model , hp):
        self.model = model
        self.engine = Engine(hp)

    def start(self):
        print(f"Car {self.model} start with {self.engine.start()}")

bm = Car(2027, 150)

bm.engine = 98787
```

```python
class Engine:
    def __init__(self, hp):
        self.hp  = hp

    def start(self):
        return f"Engine start with {self.hp}"

class Car:
    def __init__(self, model , hp):
        self.model = model
        self.engine = hp

    def start(self):
        print(f"Car {self.model} start with {self.engine.start()}")

    @property
    def engine(self):
        return self.__engine
    @engine.setter
    def engine(self, val):
        if isinstance(val, Engine):
            self.__engine = val
        elif isinstance(val, int):
            self.__engine = Engine(val)
        else:
            raise TypeError("Invalid engine type")


c = Car(2026, 100)
print(c.__dict__)
```

```python
class Notifier:
    def send(self, msg):
        raise TypeError("Noooo")

class SMS(Notifier):
    def send(self, msg):
        print(f"sending SMS {msg}")

class Email(Notifier):
    def send(self, msg):
        print(f"sending email {msg}")

for n in [SMS(), Email()]:
    n.send("Hello world")
```

```python
class Tests:
    pass

a = Tests()
b = Tests()
myset = {a , b }

print(myset) # objects are hashable

a.name = 'etet'

print(a.__dict__)

popped_a = myset.pop()
print(popped_a.__dict__)







```

<div style="background:#fff1f2;border-left:6px solid #e11d48;border-radius:14px;padding:14px 16px;margin:8px 0;font-family:system-ui,sans-serif;">
  <div style="color:#be123c;font-weight:800;font-size:15px;margin-bottom:4px;">Equality without a hash turns the set away</div>
  <div style="color:#1e293b;font-size:14.5px;line-height:1.5;">Defining <code>__eq__</code> and not <code>__hash__</code> makes Python set <code>__hash__ = None</code>. Equal objects would need a stable hash, and the identity hash would no longer match. The set refuses the object: <code>unhashable type: 'Tests'</code>.</div>
</div>


```python
class Tests:
    def __eq__(self, other):
        if isinstance(other, Tests):
            return self.__dict__ == other.__dict__

a = Tests()
b = Tests()
myset = {a , b }

print(myset) # objects are hashable

```

```python
class Scaler:
    def __init__(self, factor):
        self.factor = factor
    def __call__(self, val):
        return self.factor * val


d = Scaler(5)
print(d, d.__dict__)
print(callable(d))

d(2)
```

<div style="background:linear-gradient(90deg,#0f766e,#4f46e5);border-radius:18px;padding:18px 20px;color:white;font-family:system-ui,sans-serif;margin-top:6px;">
  <div style="font-size:12px;letter-spacing:0.16em;text-transform:uppercase;opacity:0.8;font-weight:700;">Protocol · with</div>
  <div style="font-size:24px;font-weight:800;margin-top:4px;">Enter, do the work, exit</div>
  <div style="font-size:15px;line-height:1.5;margin-top:6px;opacity:0.95;"><code>with</code> calls <code>__enter__</code> before the body and <code>__exit__</code> after it, including when the body raises. Returning <code>False</code> means “do not swallow that error.” The instance starts with an empty dictionary: this timer keeps no attributes.</div>
</div><div style="margin-left:0px;background:#d1fae5;color:#065f46;border-radius:10px;padding:7px 10px;margin-top:6px;font-family:ui-monospace,SFMono-Regular,Menlo,Consolas,monospace;font-size:13px;line-height:1.4;"><strong style="margin-right:6px;">●</strong>__enter__ prints "timer started"</div><div style="margin-left:22px;background:#d1fae5;color:#065f46;border-radius:10px;padding:7px 10px;margin-top:6px;font-family:ui-monospace,SFMono-Regular,Menlo,Consolas,monospace;font-size:13px;line-height:1.4;"><strong style="margin-right:6px;">●</strong>the body prints "My timer"</div><div style="margin-left:0px;background:#dbeafe;color:#1e3a8a;border-radius:10px;padding:7px 10px;margin-top:6px;font-family:ui-monospace,SFMono-Regular,Menlo,Consolas,monospace;font-size:13px;line-height:1.4;"><strong style="margin-right:6px;">↩</strong>__exit__ prints "timer finished"</div>


```python
class Timer:
    def __enter__(self):
        print("timer started")

    def __exit__(self, exc_type, exc_val, exc_tb) :
        print("timer finished")
        return False
```

```python
t = Timer()
print(t.__dict__)
```

```python
with t:
    print("My timer ")
```

```python
class Notifier:
    def send(self, msg): raise NotImplementedError

class SMS(Notifier):
    def send(self, msg):
        return (f"sending SMS {msg}")

class Email(Notifier):
    def send(self, msg):
        return (f"sending email {msg}")

class Alert:
    def __init__(self, notifier):
        self.notifier = notifier
    def exam_moved(self):
        return self.notifier.send("Exam is moved to xyz ")

a = Alert(Email())
a.exam_moved()
```
