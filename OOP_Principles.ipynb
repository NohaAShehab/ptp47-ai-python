```python
class Course:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self, new_price):
        if(isinstance(new_price, int)) and new_price > 0:
            self.__price = new_price
        else:
            raise TypeError("Price must be an integer")

    @price.deleter
    def price(self):
        print( "property deleted")

python = Course("Python", 100)
print(python.price)
```

```python
class Course:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self, new_price):
        if(isinstance(new_price, int)) and new_price > 0:
            self.__price = new_price
        else:
            raise TypeError("Price must be an integer")

    @price.deleter
    def price(self):
        print( "property deleted")
        del self.__price

python = Course("Python", 100)
print(python.price)
```

```python
del python.price
```

  <div style="color:#1e293b;font-size:14.5px;line-height:1.5;"><code>python.price</code> still calls the getter. The getter returns <code>self.__price</code>, and that name is gone, so Python raises <code>AttributeError: _Course__price</code>. The property did not vanish. The value behind it did.</div>
