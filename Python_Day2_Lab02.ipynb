**Worked example.** Count how many words in a list are longer than a limit.

| Step | What you write |
| --- | --- |
| Inputs | `words`: a list of strings. `limit`: an int, 3 by default. |
| Output | An int: how many words have more than `limit` letters. |
| By hand | `["AI", "python", "data"]` with limit 3 gives 2. An empty list gives 0. |
| Code and test | See the cell below. |

````python
def count_long(words, limit=3):
    """Return how many words have more than `limit` letters."""
    count = 0
    for word in words:
        if len(word) > limit:
            count += 1
    return count

print(count_long(["AI", "python", "data"]))   # 2
print(count_long([]))                         # 0
print(count_long(["AI", "python"], limit=1))  # 2
````
