```python
myt = tuple("noha")
print(myt)
```

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>tuple('noha')</code> prints <code>('n', 'o', 'h', 'a')</code>.</div>

<div style="background:#DFF3EA; border-left:6px solid #17885B; border-radius:8px; padding:10px 14px; margin:8px 0; color:#111111;"><b style="color:#17885B;">Result.</b> <code>tuple(['noha'])</code> prints <code>('noha',)</code>. The name is not spelled out.</div>
